```python
import numpy as np
import matplotlib.pyplot as plt

#Echauffement

a = np.array([1, 2, 3])

#1) a[:,None] est de format (1,3), et a[:,None] de format (3,1) (On rajoute une dimension là où est le None).
#2) a[:,None]-a[None,:] est de format (3,3). En notant a_i le i-ème élément de a, on a alors que l'élément [i,j] est a_i-a_j.

b=np.array([[1,2,3],[4,5,6]])

#3) b[:,None,:] est de format (2,1,3), et b[:,:,None] est de format (2,3,1) (en terme de dimensions, on insère un 1 à l'emplacement du None). 
#D'après les règles du broadcasting, cela implique que leur différence est de format (2,3,3). De plus, en notant b_ij l'élément sur la ligne i, colonne j de b, on a que l'élément [i,j,k] de b[:, None, :] - b[:, :, None] vaut : b_ij-b_ik. 

```

```python
#Initialisation aléatoire

# les bornes pour le tirage au sort initial
mass_max = 3.
    
x_min, x_max = -10., 10.
y_min, y_max = -10., 10.

speed_max = 1.

def init_problem(N):
    masses=np.random.uniform(0.01*mass_max,mass_max,N) #on ne met pas 0 en borne inf pour éviter des divisions par 0
    positions=np.random.uniform(np.array([[x_min], [x_max]]),np.array([[y_min], [y_max]]), (2,N)) #On souhaite avoir un tableau où les positions selon x (resp. y) sont sur la 1re ligne (resp. 2ème ligne) du tableau positions, d'où les tableaux de bornes inf et sup
    speeds=np.random.uniform(np.array([[0.01*speed_max],[speed_max]]),np.array([[0.01*speed_max],[speed_max]]), (2,N))
    return np.array(masses),positions,speeds


masses, positions, speeds = init_problem(10)

#ceci devrait afficher OK
try:
    assert masses.shape == (10,) and positions.shape == speeds.shape == (2, 10)
    print("OK")
except:
    print("KO")

```

```python
#fonction initialisation fournie

def init3():
    # first element is sun-like: heavy, at the center, and no speed
    masses = np.array([3, 1, 1], dtype=float)
    positions = np.array([
        [0, 5, -5], 
        [0, 1, -1]], dtype=float)
    speeds = np.array([
        [0, -1, 1], 
        [0, 0, 0]], dtype=float)
    return masses, positions, speeds

```

```python
#Les forces

def forces(masses,positions,G=1.0):
    vecteurs=positions[:,None,:]-positions[:, :, None] #Pour obtenir un tableau de taille (N,N,2), donnant le vecteur r_j-r_i (force attractive)
   
    distance=np.linalg.norm(vecteurs, axis=0) 
    #On calcule les distances selon la dimension possédant deux éléments, i.e. la 1re
    
    np.fill_diagonal(distance,np.inf) 
    #Pour gérer le problème de dist[i,i], on décrète que l'objet i est à distance infinie de lui-même : cela signifie que son auto-contribution est nulle, puisqu'on divise par la distance.
    
    coefs=(G*masses[:,None]*masses[None,:])/distance**3 
    #De même, on raisonne comme dans l'échauffement pour obtenir un tableau de taille (N,N)
    
    forces=np.sum(coefs[None,:,:]*vecteurs,axis=-1)
    #On somme selon l'axe des j pour obtenir un tableau forces de taille (2,N), où l'élément [1,i] (resp. [2,i]) est la force selon x (resp. y) subie par l'objet i.
    return forces




# pour tester, voici les valeurs attendues avec la config prédéfinie

masses, positions, speeds = init3()

f = forces(masses, positions)
print(f)

# should be true
np.all(np.isclose(f, np.array([
    [ 0.        , -0.12257258,  0.12257258],
    [ 0.        , -0.02451452,  0.02451452]])))





```

```python
#le simulateur

def simulate(masses,positions,speeds,dt=0.1,nb_steps=100):
    resultat=np.zeros((nb_steps,positions.shape[0],len(masses)))

    #On a dim=positions.shape[0], N=len(masses)

    for k in range(nb_steps):

        accelerations=forces(masses,positions,G=1.0)/masses[None,:] 
        #On rajoute un None de sorte que la division soit possible : les deux tableaux seront de taille (dim,N)
        #Il faut recalculer l'accélération à chaque fois, car les forces dépendent des distances
    
        speeds+=accelerations*dt #On incrémente les vitesses de l'accélération fois le pas de temps, car dv = a*dt
       
        positions+=speeds*dt #De même pour les positions avec les vitesses
        
        resultat[k,:,:]=positions #On met à jour la position k du résultat
    
    return resultat


# pour tester

SMALL_STEPS = 4

s = simulate(masses, positions, speeds, nb_steps=SMALL_STEPS)

try:
    if s.shape == (SMALL_STEPS, 2, 3):
        print("shape OK")
except Exception as exc:
    print(f"OOPS {type(exc)} {exc}")

# pour tester: should be true

positions1 = s[0] #C'est le 1er changement
print(s)

np.all(np.isclose(positions1, np.array([
    [ 0.        ,  4.89877427, -4.89877427],
     [ 0.        ,  0.99975485, -0.99975485] ])))

```

```python
#Le coloriage

def draw(simulation, masses, colors, scale=10.):
    if colors.all()==None : 
        colors=np.random.randint(0,255,size=(len(masses),3))
        #Si on n'a pas renseigné de couleurs, i.e. aucun élément de colors n'est rempli, cela en génère au hasard
    for k in range(len(simulation)):
        fig, ax = plt.subplots()
        #pour créer un nouvel environnement de dessin à chaque pas de temps
        ax.scatter(simulation[k, 0, :], simulation[k, 1, :], color=colors,s=masses*scale) #on pondère la taille par la masse
        plt.close(fig)


colors3 = np.array([
    [32, 32, 32],
    [228, 90, 146],
    [111, 0, 255],
]) / 255

masses, positions, speeds = init3()

simulation = simulate(masses, positions, speeds)

draw(simulation, masses, colors3)
```
